```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)
delta = (n == 0).astype(float)
# unit impulse
step  = (n >= 0).astype(float)
# unit step
ramp  = n * step
# unit ramp
expo  = 0.8**n * step
# real exponential
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

```python
n1 = np.arange(-5, 16)
step1 = (n1 >= 0).astype(float)

a_values = [0.5, -0.8, 1.1]
fig1, axes1 = plt.subplots(1, 3, figsize=(14, 3.5))

for ax, a in zip(axes1, a_values):
    x = (a ** n1) * step1
    ax.stem(n1, x, basefmt="k-")
    ax.set_title(f"({a})^n u(n)")
    ax.set_xlabel("n")
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
n2 = np.arange(0, 41)
x1 = np.cos(2 * np.pi * n2 / 8)
x2 = np.cos(2 * np.pi * n2 / 7.5)

fig2, axes2 = plt.subplots(2, 1, figsize=(12, 6))

axes2[0].stem(n2, x1, basefmt="k-")
axes2[0].set_title("cos(2πn / 8)  [Period N = 8]")
axes2[0].set_xlabel("n")
axes2[0].grid(True, linestyle='--', alpha=0.5)

axes2[1].stem(n2, x2, basefmt="k-")
axes2[1].set_title("cos(2πn / 7.5) [Period N = 15]")
axes2[1].set_xlabel("n")
axes2[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import os
from scipy.io import wavfile
import numpy as np
import matplotlib.pyplot as plt

fname = "bily.wav"
if os.path.exists(fname):
    # change to your file name
    fs, x = wavfile.read(fname)
    if x.ndim > 1:
        # stereo -> mono
        x = x.mean(axis=1)
    x = x / np.max(np.abs(x)) # normalise to [-1, 1]
else:
    # fallback test signal
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs
print(f"fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}")
plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.show()
```
